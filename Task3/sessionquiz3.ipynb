## Question 3.2: Calculate escape velocities

The escape velocity for a mass m from a heavy object with mass M given the distance r between these objects, is calculated with:

$\frac 12 mv^2=\frac{GMm}{r}\rightarrow v_{esc}=\sqrt{\frac{2GM}{r}}$

```python
#VERSION ONE (FINAL)
from math import sqrt
from scipy.constants import G

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']

system = zip(masses, radii, names) #"bundle" with mass, radius and name tied together

escape = {} # New empty dictionary

for mass, radii, name in system:
    escapeV = sqrt((2*G*mass)/(radii))
    escape[name] = escapeV # add to dictionary
    print(f"{name}: mass is {mass:.2e}, escape velocity is {escapeV: .4e}")
```

## Question 3.7: Angular Size

In astronomy, we often relate size and distance of objects to their angular size. If the angular size is $\delta$, the size of the object is $d$, and the distance is $D$, then the relation between these parameters is:
 
$$\delta=2\arctan(\frac{d}{2D})$$

Write a function that has input parameters $d$ and $D$, and returns the angular size $\delta$ in degrees.

```python
from math import *

def AngularSize(d, D): # Define function
    Delta_rad = 2*atan(d/(2*D))
    Delta_deg = degrees(Delta_rad)
    return Delta_deg

d = float(input("size of object: ")) # size of object d
D = float(input("distance: ")) # distance D

angsize = AngularSize(d, D) # make into a value of a function

print(angsize)
```

## Question 3.10: Area and Volume of a cylinder

Write a function with two arguments and a return value according to the recipe below:

Given a cylinder with radius $r$ and height $h$, one can derive the following expressions for the area $A$ and volume $V$:

$$A=2*\pi r^2+2\pi r h$$
$$V=\pi r^2 h$$

Create a function $\text{\tt{cylinder()}}$ which calculates and returns the **area** and **volume** of a cylinder given its radius and height.

After the function call, use the result to calculate the ratio: area/volume.


```python
from math import *

def cylinder(r, h):
    A = 2*pi*r**2+2*pi*r*h
    V = pi*r**2*h
    return A, V

r = float(input("Radius: "))
h = float(input("Height: "))

Area, Volume = cylinder(r, h)

print(f"Area is {Area:.3f} and Volume is {Volume:.3f}")

print(f"Ratio is {Area/Volume}")
```

## Question 3.17: Standard deviation using ndarray and in a one-liner

The standard deviation for numpy.std() is the two pass algorithm (i.e. it needs two loops). In formulas this is:
$$\sigma = \sqrt{\frac 1N \sum_{i=0}^{N-1} (y_i-\mu)^2}$$

where:
$$\mu = \frac 1N \sum_{i=0}^{N-1}y_i$$

- Create a function called stdev() using NumPy arrays for the aforementioned equations, which returns the standard deviation of an input array.
- Given any input array, compare your result to numpy.std()
- Change your code into a one-liner by substitution so that it gives the same output as numpy.std(). See hints below.

```python
import numpy as np
import math

y = np.array([])  # Input array
N = int(input("Enter the number of values you want:  "))

for i in range(N):
    value = float(input("Element:  "))
    y = np.append(y, value) # Add elements to array
print(y)

# mean (mu) formula
SumMu = 0
for i in range(N):
    SumMu += y[i]

mu = SumMu/N

#stdev

def stdev(y, mu):
    total =0  # sum starts at 0
    for i in range(N):
        total += (y[i] - mu)**2  # stdev formula
    Sigma = math.sqrt(total/N)
    return Sigma

standarddev = stdev(y, mu)

print(standarddev)
```

## 3.18: Least Squares fitting

Now let's try to perform a linear fit for data using NumPy array.

Assume you got a list with measured distances (Mpc) and velocities (km/s) of a galaxy from an observer. The distances are in the first column of the table and the velocity is the second column:


```python
import numpy as np

data = np.array([[18.49, 2094.75],
[35.48, 2320.87],
[35.48, 1224.9],
[11.34, 872.13],
[7.12,  432.2],
[9.82,  751],
[23.66, 1410],
[44,    2700],
[66.7,  5292],
[52.4,  2550],
[55, 5253],
[25.6,  2419],
[26.19, 2158]])

x = data[:, 0] #distances
y = data[:, 1] #velocities

print(f"Distances: {x}")
print(f"Velocities: {y}")

#Averages
avgX = np.mean(x)
avgY = np.mean(y)

# slope b
b = (np.sum((x-avgX)*(y-avgY))/(np.sum((x-avgX)**2)))

# intercept a
a = avgY - b*avgX

print(f"Slope (b): {b}")
print(f"Intercept (a): {a}")

print(f"fitted line: y = {a:.3f}+{b:.3f}x")
```

## 3.22: Plotting a catenary

```python
import numpy as np
from matplotlib import pyplot as plt

# integer x values
x = list(range(-5, 6))  # x values ranging from -5 to 6

# function
y = []
for value in x:
    y.append(np.cosh(value))

#plot
plt.figure()

plt.plot(x, y, marker='o', color='r', label='y = cosh(x)')

#more plotting

plt.title("Catenary Curve", fontsize=12)
plt.xlabel("x values", fontsize=12)
plt.ylabel("y = cosh(x)", fontsize=12)

plt.grid()
plt.legend()

plt.show()
```
