## Question 4.1: List comprehension

Create two lists x = [1,2,3,10] and y = [10, 20, 30, 40].

- Use list comprehension with function zip() to add the two lists (i.e. z[i] = x[i] + y[i] for i = 0,1,2,3).
- Can you make the code one line?
- Secondly, figure a way to use function map(), as instructed in the above notebook, to obtain the same result.


```python
X = [1, 2, 3, 10]
Y = [10, 20, 30, 40]

Z = []

for x, y in zip(X, Y):
    Z.append(x+y)

print(Z)

''' OR 
for i in range(len(X)):
    Z.append(x[i]=y[i]) '''
```

```python
X = [1, 2, 3, 10]
Y = [10, 20, 30, 40]

Z = [x+y for x, y in zip(X, Y)]

print(Z)
```

```python
def summ(x, y):
    return x+y

X = [1, 2, 3, 10]
Y = [10, 20, 30, 40]

summed = map(summ, X, Y)

print(list(summed))
```

## Question 4.3: Game making: Rock-Paper-Scissors

Build a small game for playing Rock-Paper-Scissors. You will play with the programme by typing R, P or S, and the programme will randomly generate a R, P, or S to play with you.

    Allow the user to input, R, P, or S to start playing. Give clear instruction during the input.

    The programme will generate R, P, or S per time, built on a function from module numpy.random.

    Judge the result -- who wins, who loses.

    Make appropriate comment on your code.

    Make it an executrable scripit and run on your terminal.

    This file needs to be upload to GitHub. Show your GitHub website to TA/lecturer.


```python
import numpy as np

player = str(input("Choose Rock (R), Paper (P) or Scissors (S)" ))  # Input players choice

game = np.array(['Rock', 'Paper', 'Scissors'])       # Array of choices
indx = np.random.randint(0, len(game), 1)            # Index of the choices
print(f"Opponents choice: {game[indx]}")

# If player chooses Rock (R)

if player == "R" and game[indx] == 'Rock':
    print("You TIE")
elif player == "R" and game[indx] == 'Paper':
    print("You LOSE")
elif player == "R" and game[indx] == 'Scissors':
    print("You WIN")

# If player chooses Paper (P)

elif player == "P" and game[indx] == 'Rock':
    print("You WIN")
elif player == "P" and game[indx] == 'Paper':
    print("You TIE")
elif player == "P" and game[indx] == 'Scissors':
    print("You LOSE")

# If player chooses Scissors (S)

elif player == "S" and game[indx] == 'Rock':
    print("You LOSE")
elif player == "P" and game[indx] == 'Paper':
    print("You WIN")
else:
    print("You TIE")
```
