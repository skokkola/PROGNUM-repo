- Calculate the mean and standard deviation. Are these close to the theoretical values (i.e. your input values for the distribution)?
- The calculated values for mean and standard deviation must be listed in the legend of the plot.
- Look it up -- draw a vertical line which corresponds to the sample mean.
- Look it up -- draw a shading area representing the range covered by one standard deviation w.r.t. the mean.
- Use scipy.stats.norm.cdf() to compute the probability of any values, X, that is smaller than and equal to the indicated height, h, i.e. $P(X \leq h)$
- Use scipy.stats.norm.cdf() to compute the probability of a height located within 1 standard deviation range, i.e. $\mu - \sigma \leq h \leq \mu + \sigma$


```python
import numpy as np
from matplotlib import pyplot as plt
import math
from scipy.stats import norm

lengths = np.random.normal(185, 8, 100000)

plt.hist(lengths, bins=40)

std = np.std(lengths)
mean = np.mean(lengths)

print(f"Standard deviation is {std:.2f}, which is close to given standard deviation (8 cm). \nMean is {mean:.2f}, which is close to given mean (185 cm)")

plt.axvline(mean, color='red', linestyle='dashed', linewidth=2, label=f'Mean: {mean:.2f}')
plt.axvline(mean + std, color='green', linestyle='dashed', linewidth=2, label=f'Standard deviation: {std:.2f}')
plt.axvline(mean - std, color='green', linestyle='dashed', linewidth=2)

plt.axvspan(mean - std, mean + std, alpha=0.3, color='green', label='Standard Deviation')

plt.legend()
plt.show()


cdf = norm.cdf(185, mean, std)

# prob = 1 - cdf for when we want more than value

prob = cdf * 100
print(f"Probability to find a length less than or equal to 185 cm: {prob:.3f}")

prob2 = (norm.cdf(mean + std, mean, std) - norm.cdf(mean - std, mean, std))*100
print(f"Probability to find a length within standard deviation range: {prob2:.3f}")
```

```python
import numpy as np
from numpy import sin, cos, tan, log, log10
from scipy.integrate import quad

# Start and end values

a = float(input("Enter starting value: "))
b = float(input("Enter ending value: "))

# Generate 100 x values from a to b

x = np.random.uniform(a, b, 100)

# User input function

user_function = input("Enter a function f(x)=")

# Evaluate user input into mathematical form

def function(x):
    y = eval(user_function)
    return y

Function = function(x)

n = len(x) # Number of values

Sum = 0  # Sum starts from 0

# Sums the values of the function together

for i in range(1, n):
        Sum += Function[i]

# Monte Carlo function

MC = ((b-a)/n)*Sum

print(f"Integral of function f(x)={user_function} is {MC:.3f}")
```

## Question 4.12: Rotate a square

- Create a complex rotor for angle $\theta = 30$ degrees.

- Plot the square:

x = np.array([1,3,3,1,1])
y = np.array([1,1,3,3,1])

- Use the rotor to rotate the square over angle. Note that positive angles are measured counter clockwise.

- Plot the rotated square with different colour.


```python
from matplotlib import pyplot as plt
import numpy as np
import math

x = np.array([1, 3, 3, 1, 1])
y = np.array([1, 1, 3, 3, 1])

z = x + y*1j               # (x,y) coordinates as complex numbers

angle_degs = 30
theta = math.radians(angle_degs)

# Define rotor

rotor = math.cos(theta) + math.sin(theta)*1j  

zr = rotor *z              # Rotated function
xr = zr.real               # Rotated x-values
yr = zr.imag               # Rotated y-values

# Original square
plt.plot(x, y, color='b', marker='o', label='Original square')

# Rotated square
plt.plot(xr, yr, color='r', marker='o', label='Rotated square')

plt.xlabel('Real')
plt.ylabel('Imaginary')

plt.gca().set_aspect(1)
plt.grid()
plt.show()
```

## Question 4.16: Calculating integration using quad()

Cosmological Lookback Time Calculation

In a flat ΛCDM universe, the lookback time to redshift z is:



- Call astronomical constants, $H_0, \Omega_m$, and $\Omega_{\lambda}$ from AstroPy package:
- Use scipy.integrate.quad() to compute $t_L(z)$ 

- Convert the lookback time to Gyr.

- Try $z=2$ (Answer: 10.53 Gyr).



```python
from astropy.cosmology import Planck18
from math import sqrt
from scipy import integrate

H0 = Planck18.H0.value
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0

z = 2

def Hz(z):                              # Define H(z)
    return H0*sqrt(Om0*(1+z)**3+Ode0)

def t(z):                               # Define t_L(z)
    return 1/((1+z)*Hz(z))

t_L, err = integrate.quad(t, 0, z)      # Integration

# t_L, err : Because quad() returns a tuple, and we can't use that when converting to Gyr,so we unpack it

# Mpc in km
Mpc_to_km = 3.0857e19  # km

# Second in Gyr
sec_to_Gyr = 1 / 3.1536e16  # Gyr

tL_Gyr = t_L * Mpc_to_km * sec_to_Gyr

print(f"Lookback time is {tL_Gyr:.2f} Gyr")
```

Probability Density Function (parametrized in terms of mean and the variance (std)):

$$ \large{f(x: \mu, \sigma^2)= \frac{1}{\sigma\sqrt{2\pi}}e^{-\frac 12 (\frac{x-\mu}{\sigma})^2}}$$

```python
import numpy as np
from matplotlib import pyplot as plt
from math import *

# Random distributions  (loc = mean, scale = std)

X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

# Plotting X and Y

plt.hist(X, bins=30, alpha=0.5, density=True, color='blue', edgecolor='black', lw=1, label='X')   # Edgecolor is the vertical lines
plt.hist(Y, bins=30, alpha=0.5, density=True, color='red', edgecolor='black', lw=1, label='Y')

# PDF lines won't show w/o density=True (makes histogram match the PDF scale)

# Probability Density Function

def PDF(x, mu, sigma):
    return 1/(sigma*np.sqrt(2*pi))*np.exp(-(1/2)*((x-mu)/sigma)**2)

xx = np.linspace(min(X.min(), Y.min()), max(X.max(), Y.max()), 300)   # Ordered x'-values (from PDF) instead of random values in X and Y

X_pdf = PDF(xx, 0, 1)
Y_pdf = PDF(xx, 2, 1)

# Plotting PDF

plt.plot(xx, X_pdf,dashes=[2, 4] , color='black',  lw=3, label='X(PDF)')   # lw = line width
plt.plot(xx, Y_pdf, dashes=[6, 2], color='black',  lw=3, label='Y(PDF)')

plt.xlabel('Values')
plt.ylabel('Probability density')
plt.title("1D Histogram + PDF")

plt.xlim(-6, 8)   # Max and min x values

plt.grid(axis='y', dashes=[6, 4])   # Grid lines on y-axis with dashes


# Adding text

plt.annotate(r"$\mathcal{N}(0,1)$", xy = (-1.75, 0.155), xytext=(-5, 0.15), color = "blue", fontsize=12,     # fc = fill color, ec = outline
             arrowprops=dict(arrowstyle="simple", fc="blue", ec="blue", mutation_scale=30))

plt.annotate(r"$\mathcal{N}(2,1)$", xy = (3.5, 0.155), xytext=(5, 0.15), color = "red", fontsize=12, 
             arrowprops=dict(arrowstyle="simple", fc="red", ec="red", mutation_scale=30))

plt.legend()
plt.show()
```

```python
import numpy as np
from matplotlib import pyplot as plt
from math import *

X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

# 2D Histogram 

plt.figure(figsize=(8,6))

# Create the 2D histogram (each square is a bin: 30 x 30 = 900)

plt.hist2d(X, Y, bins=30, cmap='Blues')

plt.colorbar()

plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Histogram + Contour")

# Grid for PDF

x = np.linspace(-5, 5, 200)     # Creates evenly spaced values, ranging from min to max values
y = np.linspace(-3, 7, 200)

X_grid, Y_grid = np.meshgrid(x, y)    # Creates grid, each point in grid represents a coordinate (x, y)

# 2D Gaussian PDF  (product of two 1D  PDFs) (exp decay= decreases away from center

Z = (1/(2*np.pi))*np.exp(-(1/2)*(X_grid**2 + (Y_grid-2)**2))


# Contour levels (lines of equal probability density)  (distance from mean)

levels = [
    np.exp(-0.5*(3**2))/(2*np.pi),      # 3 sigma
    np.exp(-0.5*(2**2))/(2*np.pi),      # 2 sigma
    np.exp(-0.5*(1**2))/(2*np.pi),      # 1 sigma
]

# Contour lines (X_g and Y_g are coordinates of the grid, and Z the PDF values at those coordinates)

CS = plt.contour(X_grid, Y_grid, Z, levels=levels, colors='red', lw=2)

fmt = {                           # Label contours
    levels[2]: r"$1\sigma$",
    levels[1]: r"$2\sigma$",
    levels[0]: r"$3\sigma$"
}

plt.clabel(CS, inline=True, fmt=fmt)   # Add labels to contours (inline = label inside line)

plt.xlim(-4, 4)
plt.ylim(-2, 6)

plt.grid(True, dashes=[6, 4], alpha=0.6)    # True turns grid on

plt.show()
```

```python
 import numpy as np
from matplotlib import pyplot as plt
from math import *
from scipy.stats import gaussian_kde

# Generate random data
X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

# 2D Histogram
plt.figure(figsize=(8,6))

# Create the 2D histogram (each square is a bin: 30 x 30 = 900)
plt.hist2d(X, Y, bins=30, cmap='Blues')

plt.colorbar()

plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Histogram + Contour")

# Grid for PDF
x = np.linspace(-5, 5, 200)    # Creates evenly spaced values, ranging from min to max values
y = np.linspace(-3, 7, 200)

X_grid, Y_grid = np.meshgrid(x, y)    # Creates grid, each point in grid represents a coordinate (x, y)


# kernel density

positions = np.vstack([X_grid.ravel(), Y_grid.ravel()])       # positions = creates 2-row array with original data points
data = np.vstack([X, Y])             # np.vstack = stacks arrays vertically, data = creates 2 row arrat with original data
kde = gaussian_kde(data)            # creates prob density functions
Z = kde(positions).reshape(X_grid.shape)       # evaluates density at each grid point, reshape= converts flat array to grid shape

# contour levels

sorted_z = np.sort(Z.ravel())               # sorts from lowest to highest
cdf = np.cumsum(sorted_z)/np.sum(sorted_z)  # calculates Cumulative Distribution Function
sorted_z_desc = sorted_z[::-1]              # reversed: highest to lowest
cdf_desc = 1-cdf[::-1]                      # creates complementary CDF (probability of exceeding a value)


level_1s = sorted_z_desc[np.searchsorted(cdf_desc, 0.68)]         # searchsorted = finds density threshold where 68% of probability is above (1 sigma)
level_2s = sorted_z_desc[np.searchsorted(cdf_desc, 0.95)]
level_3s = sorted_z_desc[np.searchsorted(cdf_desc, 0.997)]

levels = [level_3s, level_2s, level_1s]       # stores 

CS = plt.contour(X_grid, Y_grid, Z, levels=levels, colors='red', lw=2)


sigma_values = [3, 2, 1]
fmt = {}
for i, sigma in enumerate(sigma_values):
    fmt[levels[i]]=f'{sigma}\u03C3'

plt.clabel(CS, inline=True, fmt=fmt, fontsize=12)

plt.xlim(-4, 4)
plt.ylim(-2, 6)

plt.grid(True, dashes=[6, 4], alpha=0.6)    # True turns grid on

plt.tight_layout()
plt.show()
```
